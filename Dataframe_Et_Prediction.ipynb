**Question #4 :** Appliquez un algorithme de régression multivariée avec le contenu du fichier Final.csv pour trouver les éléments qui prédisent la valeur obtenue colonne IPM_TOTAL (4 points).

```python
import pandas as pd
import requests
from io import StringIO
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

# ID du fichier final.csv sur google drive
file_id = '1p9ME7y-Blq_c-Q-cO9p9qY-7fRxB9DS9' # Lien : https://drive.google.com/file/d/1p9ME7y-Blq_c-Q-cO9p9qY-7fRxB9DS9/view?usp=drive_link
direct_link = f'https://drive.google.com/uc?id={file_id}'

# Téléchargement du fichier final.csv
response = requests.get(direct_link)

# Lecture du contenu du fichier final.csv
data = response.content.decode('utf-8')

# Création d'un dataframe à partir des données dans le fichier final.csv
final_df = pd.read_csv(StringIO(data))

# Sépareration des fonctionnalités (X) et de la cible (y)
X = final_df.drop(columns=['ID', 'IPM_TOTAL'])  # Sélection des colonnes des fonctionnalités (tout, sauf ID et IPM_TOTAL)
y = final_df['IPM_TOTAL']  # Sélection de la colonne cible : IPM_TOTAL

# Divise les données en ensembles d'entraînement (80%) et de test (20%)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1)

# Création et entraînement du modèle de régression linéaire multivariée
model = LinearRegression()
model.fit(X_train, y_train)

# Prédictions sur l'ensemble de test
y_pred = model.predict(X_test)

# Calcul et affichage de l'erreur quadratique moyenne (MSE) pour évaluer comment le modèle a performer
mse = mean_squared_error(y_test, y_pred)
print(f"Erreur quadratique moyenne (MSE) : {mse}")

# Affichage des coefficients du modèle afin de pouvoir interpréter comment chaque fonctionnalité a contribué à la prédiction de la variable cible
print("Coefficients du modèle :")
for feature, coef in zip(X.columns, model.coef_): # Cette boucle va itérer à travers chaque paire de fonctionnalités (feature) et leur coefficient correspondant (coef)
                                                  # La fonction zip itère simultanément sur les colonnes de X (fonctionnalités) et sur les coefficients du modèle entraîné
    print(f"{feature}: {coef}")
```

Ici, j'ai programmé un algorithme d'apprentissage automatique qui prend en charge à la fois les données EEG et les données catégorielles. J'ai utilisé le réseau de neurones artificiel (MLP Classifier), car j'expérimente beaucoup avec cela dans un autre cours (INF1008) et j'ai voulu essayer d'appliquer et de pousser mes nouvelles connaissances dans ce projet.

**Question bonus #5 :**	Trouvez un algorithme d'apprentissage automatique capable de traiter à la fois les données EEG et les données catégorielles dans votre fichier final, afin de pouvoir identifier les principales colonnes prédisant la colonne IPM_TOTAL (Extra 2 points).



---



```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score
import requests
from io import StringIO

# ID du fichier final.csv sur Google Drive
file_id = '1p9ME7y-Blq_c-Q-cO9p9qY-7fRxB9DS9' # Lien : https://drive.google.com/file/d/1p9ME7y-Blq_c-Q-cO9p9qY-7fRxB9DS9/view?usp=drive_link
direct_link = f'https://drive.google.com/uc?id={file_id}'

# Téléchargement du fichier final.csv
response = requests.get(direct_link)

# Lecture du contenu du fichier final.csv
data = response.content.decode('utf-8')

# Création d'un dataframe à partir des données dans le fichier final.csv
final_df = pd.read_csv(StringIO(data))

# Définir les seuils pour les classes pour mapper les valeurs de IPM_TOTAL à des classes discrète (pour que ce soit compatible)
thresholds = [2.0, 2.5, 3.0, 3.5]

# Utilisaton de la fonction pd.cut pour mapper les valeurs de IPM_TOTAL aux classes puis ajoute une nouvelle colonne IPM_CLASS au df
final_df['IPM_CLASS'] = pd.cut(final_df['IPM_TOTAL'], bins=[-float('inf')] + thresholds + [float('inf')], labels=range(len(thresholds) + 1))

# Séparer les fonctionnalités (X) et la cible (y) du df
X = final_df.drop(columns=['ID', 'IPM_TOTAL', 'IPM_CLASS'])  # Exclure IPM_TOTAL et IPM_CLASS des fonctionnalités
y = final_df['IPM_CLASS'] # Cible = IPM_CLASS

# Diviser les données en ensembles d'entraînement (80%) et de test (20%)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1)

# Normalisation des fonctionnalités à l'aide de standardscaler
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Création et entraînement du modèle MLPClassifier. Deux couches cachés (La première = 100 neurones, deuxième = 50). Apprentissage 0.001
model = MLPClassifier(hidden_layer_sizes=(100, 50), alpha=0.001, learning_rate_init=0.001, max_iter=2000, random_state=44)

model.fit(X_train_scaled, y_train)

# Fait des prédictions sur l'ensemble de test effectué
y_pred = model.predict(X_test_scaled)

# Pour calculer la précision pour évaluer les performances du modèle
accuracy = accuracy_score(y_test, y_pred)
print(f"Précision : {accuracy}")

# Obtenir la prédiction d'influence des caractéristiques (influence de chaque colonne)
feature_weights = model.coefs_[0]  # Première couche cachée
feature_names = X.columns # Savoir quel influence correspond à quelle caractéristique
feature_importance = pd.Series(feature_weights.mean(axis=1), index=feature_names).sort_values(ascending=False) # Importance de chaque caractéristique, triée par ordre décroissant d'importance
print("Influence de chaque colonne :")
print(feature_importance)
```
